## Part 1

Suppose we had a linear regression model with one feature that learned the following bias and weights:

b=1<br>
w1=3

Further, suppose the following feature values for a given example:

x1=0

Compute the linear regression prediction for this particular example using python in Google Colab. Make sure to plot the solution.Provide an analysis for the output.

```python
import matplotlib.pyplot as plt
import numpy as np

b = 1
w1 = 3
x1 = 0

y_pred = b + w1 * x1
print(f"Prediction for x1 = {x1}: {y_pred}")

x_vals = np.linspace(-2, 2, 100)
y_vals = b + w1 * x_vals

plt.figure(figsize=(8, 5))
plt.plot(x_vals, y_vals, color='blue', label='Regression line: y = 1 + 3x')
plt.scatter(x1, y_pred, color='red', s=100, label=f'Prediction point (x1={x1}, y={y_pred})')
plt.xlabel('x₁')
plt.ylabel('Predicted y')
plt.title('Linear Regression with One Feature')
plt.grid(True)
plt.legend()
plt.show()
```

Analysis

The prediction is calculated as y = 1 + 3 × 0 = 1. The blue line in the plot represents the regression line y = 1 + 3x, which has a slope of 3 and an intercept of 1. When the feature value x₁ = 0, the product w₁ · x₁ becomes zero, so the prediction equals the bias term. The red dot marks this exact point on the line, confirming that the prediction lies on the regression line.

Analysis

The prediction for Example 1 is 1 + 9 + 2 + 8 = 20. For Example 2, it is 1 + 30 + 0 – 40 = –9. In Example 1, the negative weight w₂ and negative x₂ combine to add a positive contribution. In Example 2, the large negative term from w₃·x₃ (–40) outweighs the positive part. Each
