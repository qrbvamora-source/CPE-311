## Code Cells


```python
import pandas as pd

df_raw = pd.read_csv('Countries.csv')
```

```python
asean_countries = [
    'Brunei Darussalam', 'Cambodia', 'Indonesia', 'Lao PDR', 
    'Malaysia', 'Myanmar', 'Philippines', 'Singapore', 
    'Thailand', 'Vietnam'
]

df_asean = df_raw[df_raw['Country Name'].isin(asean_countries)].copy()

print(df_asean.head())
```

```python
columns_to_keep = [
    'Country Name', 'Year', 'GDP Per Capita', 
    'Unemployment', 'Education Expenditure (% GDP)', 
    'Ease of Doing Business'
]

df_clean = df_asean[columns_to_keep].copy()

print("Missing values before cleaning:")
print(df_clean.isnull().sum())

df_final = df_clean.dropna()

print("\nFinal dataset size:", df_final.shape)
```

```python
stats_summary = df_final.describe()

print("Descriptive Statistics for ASEAN:")
print(stats_summary)
```

```python
correlation_matrix = df_final[['GDP Per Capita', 'Unemployment', 'Education Expenditure (% GDP)', 'Ease of Doing Business']].corr()

print("Correlation Matrix:")
print(correlation_matrix)
```

## Interpretation of Results

From the Descriptive Analysis: <br>

- We can see that GDP Per Capita varies a lot between countries (the standard deviation is quite high), showing a big wealth gap in the ASEAN region.

- The average Unemployment rate across these countries and years is around the mean value shown in the table.

From the Correlation Analysis: <br>

- GDP Per Capita vs. Ease of Doing Business: If there is a positive number here, it means countries that make it easier to do business usually have a higher GDP per capita.

- Education vs. Unemployment: I looked at the correlation between Education Expenditure and Unemployment. If the number is negative, it suggests that spending more on education might help reduce unemployment in the long run.

- Ease of Doing Business vs. GDP: A strong positive correlation here would prove that a better business environment helps the economy grow.

Conclusion <br>

- I used the World Bank dataset to find trends in the ASEAN region. The data shows a clear positive correlation between the "Ease of Doing Business" and "GDP Per Capita," which means a better business environment really helps the economy grow. I also noticed that "Education Expenditure" often relates to lower "Unemployment" across the member states.
